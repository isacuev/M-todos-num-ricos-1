#**METODO** **DE** **BISECCION**



```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
def funcion1(x):
  return np.tan(x)-x
```

```python
f=funcion1(0)
```

```python
print(f)
```

```python
#INGRESAR LOS DATOS DE ENTRADA PARA LOS DIFERENTES METODOS A TRABAJAR
a=4
b=4.5

#GUARDAMOS LOS VALORES INICIALES
a0=a
b0=b

#GUARDAR VALORES INICIALESDEL ERROR Y DEL NUMERO DE ITERACCIONES
tol=0.001     #float(input("INGRESE EL VALOR DE LA TOLERANCIA: "))
nmax=100      #float(input("INGRESEEL NUMERO MAXIMO DE ITERACCIONES"))
error=100
niter=0
```

```python
#METODO DE BISECCION

#EVALUAMOS PRIMER VALOR MEDIO
m=a+(b-a)/2

#Evaluacion de la funcion en los puntos a,b y m
fa=funcion1(a)
fb=funcion1(b)
fm=funcion1(m)
```

```python
print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a0,fa,b0,fb,m,fm,error))

#CICLO ITERACTIVO
while error > tol and niter < nmax:
  m= a+(b-a)/2
  if np.sign(fa)== np.sign(fm):
    a=m
    fa=funcion1(a)
  else:
    b=m
    fb=funcion1(b)

  m=a+(b-a)/2
  fm=funcion1(m)
  error=abs(b-a)
  niter+=1

  print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a,fa,b,fb,m,fm,error))

print("La raiz de la funcion dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
