**<font color="tan"> - </font>** Consiste en generar una sucesion <font color="sienna"> ${P_n}$ </font> de aproximaciones a la raiz <font color="sienna"> $P$ </font> de la ecuacion <font color="sienna"> $f(x)=0$ </font> asi comenzando con una aproximacion inicial de <font color="sienna"> $P_0$ </font> y usando la ecuacion iterativa la cual es: <font color="sienna"> $$ P_n= P_n-1 - \frac{f(P_n-1)}{f´(P_n-1)}$$ </font> con <font color="sienna"> $n>=1$ </font> y <font color="sienna"> $f'(x)$ </font> distinto de cero.

**<font color="Wheat"> >></font>** **<font color="tan"> EJERCICIO 1 </font>** **<font color="Wheat"><< </font>**

**<font color="tan"> - </font>** Considere la funcion <font color="sienna"> $f(x)= cos(x)-x =0$ </font> y aproxime una raiz de <font color="sienna"> $f$ </font> usando el metodo de newton-raphson

**<font color="Wheat">------------------------------------------</font>** ***<font color="tan"> CODIGO </font>*** **<font color="Wheat">------------------------------------------ </font>**

**<font color="sienna">1- </font>** Importamos la libreria numpy que esta nos ayudara ya que contiene funciones matematicas y tambien agregamos matplotlib que nos ayudara a realizar la grafica

```python
import numpy as np
import matplotlib.pyplot as plt
```

**<font color="sienna">2- </font>** Definimos nustra funcion deque nos da el ejercicio

```python
def f(x):
    return np.cos(x)-x
```

**<font color="sienna">3- </font>** Ahora necesitamos determinar un intervalo donde exista una raiz para nuestra funcion, es decir:
**<font color="sienna">$$ f(a)*f(b)< 0 $$</font>**

por lo tanto tomamos el intervalo **<font color="sienna">$ [0 , \pi/2] $</font>** y evaluamos en **<font color="sienna">$ f(x) $</font>**

```python
a=0
b=np.pi/2
```

```python
print ("f(0)= ", f(a))
print ("f(pi/2)= ", f(b))
```

**<font color="tan"> - </font>** por lo tanto podemos ver que **<font color="sienna"> SI EXISTE UNA RAIZ </font>** en nuestro intervalo, ya que al evaluarlo existen signos opuestos

**<font color="Wheat">------------------------------------------</font>** ***<font color="tan"> GRAFICA </font>*** **<font color="Wheat">------------------------------------------ </font>**

```python
xx=np.linspace(a,b,100)
plt.plot(xx,f(xx))
plt.axhline(y=0, color="sienna")
```

**<font color="sienna">4- </font>** Para implementar la ecuacion iterativa tambien necesitamos definir la derivada de nuestra **<font color="sienna">$ f(x) $</font>** 

```python
def df(x):
    return -np.sin(x)-1
```

**<font color="sienna">5- </font>** Ahora determinaremos:

```python
#AGREGAMOS LISTAS PARA GUARDAR LOS RESULTADOS
lista=[]
error=[]

p0=np.pi/4  #APROXIMACION INICIAL
tol=0.0000000001   #TOLERANCIOA
nmax=100  #NUMERO MAXIMO DE ITERACCIONES 
niter=1   #CONTADOR DE ITER
```

**<font color="tan"> - </font>** En este caso determinamos nuestra aproximacion inicial mediante el **<font color="sienna"> PUNTO MEDIO DEL INTERVALO </font>** por lo tanto: **<font color="sienna">$$ P_0= \frac {0+\pi/2}{2} = \frac {\pi}{4}$$ </font>**

```python
#IMPRIMIMOS EL ENCABEZADO DE NUESTRA TABLA
print()
print("#ITER\t\t Pn-1\t\t Pn\t\t error")

#CREAMOS UN CICLO QUE MIENTRAS LAS ITERACCIONES SEAN MENOR O IGUAL AL NUMERO MAXIMO DE ITERACCIONES ESTE MISMO SIGUE
while niter<=nmax: 
    fp0=f(p0)   #EVALUAMOS TANTO LA FUNCION COMO SU DERIVADA EN NUESTRA APROXIMACION INICIAL
    dif=df(p0)

    if dif==0:
        print ("ERROR LA DERVIDA ES 0")
        break
        
    p=p0-(fp0/dif)  #APLICAMOS NUESTRA ECUACION ITERATIVA
    lista.append(p)
    
    e=np.abs(p-p0)  #CALCULAMOS EL ERROR |Pn-Pn-1|
    error.append(e)

#MOSTRAMOS LOS RESULTADOS EN NUESTRA TABLA
    print("{0} \t\t {1:.10f} \t {2:.10f} \t {3:.10f}".format(niter, p0, p, e)) 

    if e<tol: #CONDICIONAMOS QUE SI EL ERROR ES MENOR A LA TOLERANCIA PARAMOS
        print()
        print ("LA RAIZ APROXIMADA ES: ",p)
        break
        
    p0=p #ACTUALIZAMOS EL VALOR ANTERIOR CON LA NUEVA APROXIMACION
    niter+=1

else:
    print ("SE ALCANZO EL MAXIMO DE ITERACCIONES")
```
