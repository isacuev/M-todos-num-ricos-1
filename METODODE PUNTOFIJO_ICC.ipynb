**<font color="indianred">-DEFINICION </font>**

Un punto fijo para una funcion <font color="LIGHTCORAL">$g$</font>  es un numero <font color="LIGHTCORAL">$p$</font> en el que el valor de la funcion no cambia cuando se aplica la funcion <font color="LIGHTCORAL">$g(p)=p$</font> 

**<font color="indianred">-SOLUCION </font>**

**<font color="indianred">1-</font>** Ahora este metodo numerico nos va a permitir aproximar la solucion de nuestra ecuacion, para ello debemos de partir de:
<font color="LIGHTCORAL">$$f(x)=0$$</font> y mediante operaciones algebraicas debemos despejar para <font color="LIGHTCORAL">$x$</font>  siendo asi <font color="LIGHTCORAL">$$x=g(x)$$</font> 

**<font color="indianred">2-</font>** Despues debemos utilizar una aproximacion inicial  <font color="lightcoral">$p_0$</font>  y calculamos cada nueva aroximacion con <font color="lightcoral">$$p_n=g(p_n-1)$$</font>

**<font color="indianred">3-</font>** Este proceso continua hasta que el <font color="lightcoral">error</font> entre dos aproximaciones consecutivas sea menor que la toleracia que establecimos <font color="lightcoral">$${error=|p-p_0}|$$</font>

**<font color="indianred">-EJERCICIO 1 </font>**

Queremos encontrar que <font color="lightcoral">${f(x)=x^3 + 4x^2 - 10 = 0}$</font> tiene una raiz unica en <font color="lightcoral">${[1,2]}$</font>

**<font color="indianred">>>></font>** Podemos ver que al aplicar el paso numero **<font color="lightcoral">$1$</font>** de nuestra solucion, podemos ver que existen muchas formas de cambiar la ecuacion para la forma de punto fijo, en este caso tenemos las siguientes:

**<font color="indianred">A-</font>**  <font color="lightcoral">${x=g1(x)=x-x^3 - 4x^2 +10}$</font>

**<font color="indianred">B-</font>**  <font color="lightcoral">$x=g2(x)=\left( \frac{10}{x} - 4x \right)^{1/2}$</font>

**<font color="indianred">C-</font>**  <font color="lightcoral">$x=g3(x)=\frac{1}{2}(10 - x^3)^{1/2}$</font>

**<font color="indianred">D-</font>**  <font color="lightcoral">$x=g4(x)=\left( \frac{10}{4 + x} \right)^{1/2}$</font>

**<font color="indianred">D-</font>**  <font color="lightcoral">$x=g5(x)=x - \frac{x^3 + 4x^2 - 10}{3x^2 + 8x}$</font>

**<font color="indianred">2)</font>** Definimos la funcion $\color{lightcoral}{f(x)}$

```python
def funcion1(x):
  return x**3 + 4*x**2 - 10
```

**<font color="indianred">-</font>** determinamos el intervalo en nuestro codigo siendo asi:

```python
a=1
b=2
```

**<font color="indianred">3)</font>** Ahora verificamos que nuestra funcion $\color{lightcoral}{f(x)}$ evaluada en los extremos del intervalo $\color{lightcoral}{[1,2]}$ iene signos opuestos para verificar que exista una raiz, es decir: $$\color{lightcoral}{f(a) * f(b) < 0} $$

**<font color="indianred">4)</font>** Definimos las cinco distintas formas de cambiar la ecuacion para la forma de punto fijo de $\color{lightcoral}{x=g(x)}$ asi como la derivada de cada una.

```python
#PARA A

def g1(x):
    return x-x**3-4*x**2+10
def dg1(x):
    return 1-3*x**2-8*x 
```

```python
#PARA B

def g2(x):
    return np.sqrt(10/x-4*x)
def dg2(x):
    return (-10/x**2-4)/(2*np.sqrt(10/x-4*x))
```

```python
#PARA C

def g3(x):
    return 0.5*np.sqrt(10-x**3)
def dg3(x):
    return -3*x**2/(4*np.sqrt(10-x**3))
```

```python
#PARA D

def g4(x):
    return np.sqrt(10/(x+4))

def dg4(x):
    return -5/(x+4)**(3/2)
```

```python
#PARA E

def g5(x):
    return x-(x**3+4*x**2-10)/(3*x**2+8*x)

def dg5(x):
    return ((x**3+4*x**2-10)*(6*x+8))/(3*x**2+8*x)**2
    
```

```python
p0=1.5   #APROXIMACION INICIAL
tol=0.0000000001   #TOLERANCIA
nmax=30   #NUMERO MAXIMO DE ITERACCIONES 
niter=1   #CONTADOR DE ITERACCIONES
```

**<font color="indianred">6)</font>** ***<font color="indianred">TEOREMA</font>***

**<font color="indianred">>>></font>** Ahora para decidir si las funciones <font color="lightcoral">$g(x)$</font> son una buena opcion para resolver aplicaremos el ***<font color="indianred">teorema del punto fijo</font>*** que dice:

Sea <font color="lightcoral">$g \in [a,b]$</font> tal que <font color="lightcoral">$g(x) \in [a,b]$</font> para toda <font color="lightcoral">$x \in [a,b]$</font>. Ademas supongamos existen <font color="lightcoral">$g´ \in (a,b)$</font> y una constante <font color="lightcoral">$k<1$</font> con <font color="lightcoral">$$|g´(x)|<=k$$</font>  para toda <font color="lightcoral">$$x \in (a,b)$$</font> Entonces para cualquier numero <font color="lightcoral">$p_0$</font> en <font color="lightcoral">$[a,b]$</font> la sucesion definida por <font color="lightcoral">$$p_n=g(p_n-1), n>=1$$</font> Converge en el unico punto fijo <font color="lightcoral">$p$</font> en <font color="lightcoral">$[a,b]$</font>

**<font color="indianred">>>></font>** Por lo tanto vamos a verificar quien si cumplen las condiciones de nuestro teorema 

**<font color="indianred">-PARA A</font>**

```python
#CONDICION 1
#EVALUAMOS AMBOS PUNTOS DEL INTERVALO PARA VER QUE G(X) PERTENECE A [A,B]

print ("g1(1)= ", g1(a))
print ("g1(2)= ", g1(b))
print()

if g1(a)>=a and g1(b)<=b:
    print ("g1(x) SI PERTENECE DENTRO DEL INTERVALO [1,2]")
else:
    print ("g1(x) NO PERTENECE DENTRO DEL INTERVALO [1,2]")
```

```python
#CONDICION 2
# DONDE K<1 ASI |g1´(x)| <= k < 1

k=abs(dg1(a))
print ("|g1´(1)|=",k)

if k<1:
    print ("|g1´(x)| <= k < 1 , EL TEOREMA SI SE CUMPLE")
else:
    print ("EL TEOREMA NO SE CUMPLE")
```

**<font color="indianred">TOMAREMOS DE LA FUNCION D</font>**

```python
#IMPRIMIMOS EL ENCABEZADO DE LA TABLA Y LA APROXIMACION INICIAL EN LA POSICION 0 DE NUESTRA ITERACCION
print("# iter\t\t p\t\t g4(p)\t\t error")
print(f"0\t\t {p0:.10f}") 

for i in range(nmax):     
      p=g4(p0)  
      error=abs(p-p0)

#GUARDAMOS APROXIMACIONES Y ERRORES EN LAS LISTAS
      lista.append(p)
      errores.append(error)

#MOSTRAMOS RESULTADOS
      print("{0} \t\t {1:.10f} \t {2:.10f} \t {3:.10f}".format(niter, p, p0, error))

      if error < tol:
          print()
          print("LA SOLUCION APROXIMADA EN g4(X) ES: ",p)
          break
          
      p0=p #ACTUALIZAMOS EL VALOR ANTERIOR CON LA NUEVA APROXIMACION
    
      niter=niter+1 
```
