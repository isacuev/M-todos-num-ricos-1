```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 3*(x**2)-np.exp(x)
```

```python
xx=np.linspace(0,1,100)
plt.plot(xx,f(xx))
plt.axhline(y=0, color="red")
```

```python
#definimos la derivada
def df(x):
    return 6*x-np.exp(x)
```

```python
df(1)
```

```python
lista=[]
error=[]
p0=0.7

for i in range(10):
    fp0=f(p0)
    dif=df(p0)

    p=p0-(fp0/dif)
    lista.append(p)
    e=np.abs(p-p0)
    error.append(e)
    p0=p
```
