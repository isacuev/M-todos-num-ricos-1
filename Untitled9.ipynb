```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import *
```

```python
a=([0.7651977,-0.4837057,-0.1087339,0.0658784,0.0018251])
xi=([1, 1.3, 1.6, 1.9, 2.2])
```

```python
x=symbols('x')
i=symbols('i')
```

```python
p=Function('p')
```

```python
p = 0
for k in range (len(a)):
  p=p+(a[k])*product(x-(1+i*0.3),(i,0,k))
```

```python
p
```

```python
simplify(p)
```
