***<font color="purple">¿QUE ES?</font>***

El metodo de biseccion es el encontrar una raiz o solucion para una ecuacion de la forma <font color="purple">$f(x)=0$</font> para una funcion de <font color="purple">$f$</font> dada, este metodo nos realiza de manera repetida una reduccion a la mitad de los subintervalos <font color="purple">$[a,b]$</font> y luegop localizar la mitad que contiene *<font color="purple">$p$</font>*

Por lo tanto aproximaremos la solucion, calculando el punto medio del intervalo con: $$ \color{purple}{M_n = \frac {b_n-a_n}{2}}$$


***<font color="purple">EJERCICIO</font>***

En este ejercicio queremos resolver la ecuacion $\color{purple}{f(x)=x^3 + 4x^2 - 10 = 0} $ en el intervalo $\color{purple}{[1,2]}$, utilizando el metodo de biseccion para determinar una aproximacion para la raiz que sea precisa dentro de $\color{purple}{10^{-4}}$

**<font color="purple">1-</font>** Importamos la libreria numpy, que tiene funciones matematicas y matplotlib que nos permite graficar.

```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

**<font color="purple">2-</font>** Definimos la funcion $\color{purple}{f(x)}$ asociada a la ecuacion $\color{purple}{f(x)=0}$

```python
def funcion1(x):
  return x**3 + 4*x**2 - 10
```

Ahora para poder para aplicar el metodo de biseccion verificamos que nuestra funcion $\color{purple}{f(x)}$ evaluada en los extremos opuestos del intervalo $\color{purple}[a,b]$ tiene signos opuestos para verificar que exista una raiz, es decir: $$\color{purple}{f(a) * f(b) < 0} $$

**<font color="purple">3-</font>** Evaluamos nuestra funcion $\color{purple}{f(x)}$ para $\color{purple}{x=1}$

```python
f=funcion1(1)
```

```python
print(f)
```

Ahora evaluamos nuestra funcion $\color{purple}{f(x)}$ para $\color{purple}{x=2}$

```python
f=funcion1(2)
```

```python
print (f)
```

Por lo tanto podemos rectificar que **<font color="purple">SI EXISTE UNA RAIZ </font>** en nuestro intervalo **<font color="purple"> $[1,2]$ </font>**

Ahora sea para $\color{purple}{f \in C[a,b]}$ con $\color{purple}{f(a)}$ y $\color{purple}{f(b)}$ de signos opuestos, el teorema del valor intermedio asegura que existe un numero $\color{purple}{p \in [a,b]}$ tal que $\color{purple}{f(p)=0}$.

```python
#AHORA INGRESAMOS LOS DATOS DE ENTRADA DEL INTERVALO [a,b]

a=1
b=2

#GUARDAMOS LOS VALORES INICIALES DEL INTERVALO
a0=a
b0=b

tol=0.0001  #DEFINIMOS LA TOLERANCIA QUE EN NUESTRO CASO ES 10^-4
nmax=100    #ES NUESTRO NUMERO MAXIMO DE ITERACCIONES 
error=100   #EL ERROR INICIAL
niter=1     #NUESTRO CONTADOR DE ITERACCIONES 
```

**<font color="purple">4-</font>** APLICAMOS EL METODO DE BISECCION.

```python
#EVALUAMOS EL PRIMER VALOR MEDIO DE NUESTRO INTERVALO [1,2]
m=a+(b-a)/2 

#EVALUAMOS LA FUNCION EN LOS PUNTOS a,b y m
fa=funcion1(a)
fb=funcion1(b)
fm=funcion1(m)
```

**<font color="purple">5-</font>** ELABORACION DE TABLA. 

```python
#IMPRIMIMOS EL ENCABEZADO DE LA TABLA 
print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")

#MOSTRAMOS LOS PRIMEROS VALORES 
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a0,fa,b0,fb,m,fm,error))

#CICLO ITERACTIVO
while error > tol and niter < nmax:  #HAY UN CICLO MIENTRAS EL ERROR SEA MAYOR QUE LA TOLERANCIA  Y NO SOBREPASAR NUESTRO NUMERO DE ITERACCIONES 
  m= a+(b-a)/2

#COMPARAMOS F(a) CON F(b) YA QUE CON ELLO SE DECIDIRA QUE MITAD DEL INTERVALO CONSERVAREMOS 
  if np.sign(fa)== np.sign(fm):
    a=m
    fa=funcion1(a)
  else:
    b=m
    fb=funcion1(b)

  m=a+(b-a)/2
  fm=funcion1(m) 
  error=abs(b-a)
  niter+=1 

#MOSTRAMOS RESULTADOS 
  print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a,fa,b,fb,m,fm,error))

#MOSTRAMOS LA APROXIMACION FINAL DE LA RAIZ
print("La raiz de la funcion dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
