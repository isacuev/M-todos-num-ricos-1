```python
import numpy as np
import matplotlib.pyplot as plt
```

vamos a definir la funcion $f(x) = x - 3^{-x}$


```python
def f(x):
  return x-3**(-x)
```

```python
f(1)
```

```python
xx=np.linspace(-5,100,1000)
plt.plot(xx,f(xx))
plt.show()
```
